```python
# 重复值检查
dup_rows = df.duplicated().sum()
dup_customer_ids = df['customer_id'].duplicated().sum()

print(f"完全重复的行数: {dup_rows}")
print(f"重复的 customer_id 数: {dup_customer_ids}")
```

```python
missing=df.isnull().sum()
```

```python
missing_pct=(missing/len(df))*100
```

```python
missing_info=pd.DataFrame({'缺失数':missing,'缺失比例%':missing_pct})
```

```python
print("===缺失情况===")
print(missing_info[missing_info['缺失数']>0])
```

```python
print("\n=== 数据类型 ===")
print(df.dtypes)
```

```python
print("\n=== 数值列描述统计 ===")
print(df.describe())
```

```python
#日期列类型转换
date_cols=['signup_date', 'first_order_date', 'last_order_date','first_session_date', 'last_session_date']
for col in date_col:
    df[col]=pd.to_datetime(df[col],errors='coerce')
```

```python
# 验证转换结果
print("转换后的日期列类型：")
print(df[date_cols].dtypes)
```

```python
no_order=df[df['total_orders']==0]
print(f"total_order=0的用户数:{len(no_order)}")
print(f"与缺失first_order_date的行数一致？ {len(no_order)==missing['first_order_date']}")
```

```python
cat_cols_to_check = ['country', 'age_group', 'clv_tier', 'preferred_payment',
                     'preferred_device_ord', 'preferred_source', 'top_category_bought']

for col in cat_cols_to_check: 
    print(f"\n==={col}===")
    print(df[col].value_counts(dropna=False).head(10))
```

```python
# 清洗收尾：导出干净数据
df.to_csv('01_raw_data/ecommerce_data_clean.csv', index=False)
print("干净数据已导出为 ecommerce_data_clean.csv")
```
